# Sentiment Analysis with a Pre-trained Model

Let's start with a classic NLP task: sentiment analysis. We'll use a pre-trained model to classify the sentiment of a given text.

````python
from transformers import pipeline

sentiment_analyzer = pipeline('sentiment-analysis')

text_positive = "I love the new design of the website! It's so user-friendly."
text_negative = "I'm really disappointed with the customer service. They were not helpful at all."

print(f"Positive text sentiment: {sentiment_analyzer(text_positive)}")
print(f"Negative text sentiment: {sentiment_analyzer(text_negative)}")
````

# Text Generation with a Pre-trained Model

LLMs are also incredibly powerful for generating text. Let's see how we can use a pre-trained model to generate text based on a given prompt.

````python
from transformers import pipeline

text_generator = pipeline('text-generation')

prompt = "In a world where AI is becoming increasingly powerful,"
generated_text = text_generator(prompt, max_length=50, num_return_sequences=1)

print(generated_text[0]['generated_text'])
````

# Question Answering with a Pre-trained Model

Another common NLP task is question answering. We can use a pre-trained model to answer questions based on a given context.

````python
from transformers import pipeline

qa_pipeline = pipeline('question-answering')

context = """
The Hugging Face Transformers library provides a unified interface to a vast collection of pre-trained LLMs. 
It makes it easy to download, load, and use these models for a wide range of NLP tasks.
"""

question = "What does the Hugging Face Transformers library provide?"

result = qa_pipeline(question=question, context=context)

print(f"Answer: {result['answer']}")
````
