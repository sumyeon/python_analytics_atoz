# Loading and Preprocessing the Data

For this example, we'll use a sentiment analysis dataset. We'll load the data, tokenize it, and prepare it for training.

````python
from datasets import load_dataset
from transformers import AutoTokenizer

dataset = load_dataset('imdb')
tokenizer = AutoTokenizer.from_pretrained('distilbert-base-uncased')

def preprocess_function(examples):
    return tokenizer(examples['text'], truncation=True, padding='max_length')

tokenized_datasets = dataset.map(preprocess_function, batched=True)

train_dataset = tokenized_datasets['train'].shuffle(seed=42).select(range(1000))
eval_dataset = tokenized_datasets['test'].shuffle(seed=42).select(range(1000))
````

<hr>

# Fine-Tuning the Model

Now, we'll load a pre-trained model and fine-tune it on our dataset. We'll use the `Trainer` API from Hugging Face, which simplifies the training process.

````python
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer

model = AutoModelForSequenceClassification.from_pretrained('distilbert-base-uncased', num_labels=2)

training_args = TrainingArguments(
    output_dir='./results',
    eval_strategy='epoch',
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
)

trainer.train()
````

<hr>

# Evaluating the Fine-Tuned Model

After training, we'll evaluate our fine-tuned model on the test set to see how well it performs.

````python
results = trainer.evaluate()
print(results)
````
