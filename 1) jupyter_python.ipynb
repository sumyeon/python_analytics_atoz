# Key Jupyter Lab Commands

## Essential Commands 

<b> ESC   </b>   : Command mode <br>
<b> Enter </b>   : Cell edit mode <br>
<b>Shift+Enter</b> : Execute the selected cell + Move to the next cell </br>
<b>Ctrl+Enter</b>  : Execute the selected cell  <br>
<b>B, A</b>        : Insert a new empty cell below (B) or above (A) the current cell

## Frequently Used Commands (in Command mode only)

<b> C, X, V </b>     : Copy (C), Cut (X), Paste (V) the selected cell <br>
<b> Up/Down </b>    : Move to the up/down cell <br>
<b> H, K </b>        : Same as Up,Down <br>

## Helpful Commands (in Edit mode only)

<b> R, M, Y  </b>    : Set the selected cell as 'Raw', 'Markdown', 'Code' type <br>
<b> Tab </b>         : code completion hint <br>
<b> Shift+Tab </b>   : show detailed description on the current func/class/etc. <br> 
<b> Ctrl+[,]  </b>   : Indent, Unindent selected block <br>

## Functions

### Use the `def` keyword to declare a function. The arguments are listed in parentheses.

````python
def add(arg1, arg2):
    return arg1+arg2
add(1,1)
````

### You can set default values for arguments. If an argument is not provided, the default value is used.

````python
def add_n(arg1, arg2_with_default_value=10):
    return arg1+arg2_with_default_value
print(add_n(1, 1))
print(add_n(1))
````

<div class="alert alert-block alert-info">
<b> Tip:</b> Using default arguments is a great way to create flexible functions. Users can provide only the essential arguments, while advanced users can customize the function's behavior by providing additional arguments.
</div>

### A <b> Lambda </b> function is a small anonymous function.

````python
add2 = (lambda a, b: a+b)
add2(1, 1)
````

````python
(lambda a, b: a+b)(1,1)
````

<div class="alert alert-block alert-info">
<b> Tip:</b> Lambda functions are convenient for short, simple operations. They are not essential, but they can help you write more concise code.
</div>
